```python
from dataclasses import dataclass
from dtypes import DHparam, ChainParams, JointVec, Vec3
import sympy as sym
import numpy as np

def dh_transform(link: DHparam, q_i: float) -> np.ndarray:
    """Compute the Denavit-Hartenberg transformation matrix."""
    
    if link.joint_type == 'revolute':
        theta = link.theta_offset + q_i
        d = link.d
    elif link.joint_type == 'prismatic':
        d = link.d + q_i
        theta = link.theta_offset
    elif link.joint_type == 'fixed':
        theta = link.theta_offset
        d = link.d
    else:
        raise ValueError(f"Unknown joint type: {link.joint_type}")

    a = link.a
    alpha = link.alpha

    ca, sa = sym.cos(alpha), sym.sin(alpha)
    ct, st = sym.cos(theta), sym.sin(theta)

    return sym.Matrix([
        [ct, -st * ca,  st * sa, a * ct],
        [st,  ct * ca, -ct * sa, a * st],
        [0,      sa,      ca,      d   ],
        [0,      0,       0,      1   ]])
```
